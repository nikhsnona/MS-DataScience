```python
# Step 1: Import necessary libraries
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Step 2: Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Step 3: Display first few rows of the dataset
print("First few rows of Titanic dataset:")
print(titanic.head())
```

```python
# Step 4: Check for missing values
print("\nMissing values in the dataset:")
print(titanic.isnull().sum())

# Step 5: Handle missing data
# For numerical columns, we'll impute missing values with the mean.
# For categorical columns, we'll impute missing values with the mode (most frequent value).

# Separate numeric and categorical columns
numeric_columns = titanic.select_dtypes(include=['float64', 'int64']).columns
categorical_columns = titanic.select_dtypes(include=['object']).columns

# Impute missing numerical values with the mean
num_imputer = SimpleImputer(strategy='mean')
titanic[numeric_columns] = num_imputer.fit_transform(titanic[numeric_columns])

# Impute missing categorical values with the mode (most frequent)
cat_imputer = SimpleImputer(strategy='most_frequent')
titanic[categorical_columns] = cat_imputer.fit_transform(titanic[categorical_columns])

# Step 6: Check if there are any remaining missing values
print("\nMissing values after imputation:")
print(titanic.isnull().sum())
```

```python
# Step 7: Feature scaling (Standardize numerical features)
# We will standardize the numeric columns to have a mean of 0 and a standard deviation of 1.

scaler = StandardScaler()

# We will scale only the numeric columns
titanic[numeric_columns] = scaler.fit_transform(titanic[numeric_columns])

# Step 8: Display the dataset after scaling
print("\nDataset after scaling numerical features:")
print(titanic.head())
```
